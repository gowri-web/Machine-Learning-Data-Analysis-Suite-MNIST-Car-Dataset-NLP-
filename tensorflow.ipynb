```python
from tensorflow.keras.datasets import mnist
```

```python
(xtrain,ytrain), (xtest,ytest) = mnist.load_data()
```

```python
#normalization-dividing the image with the highet value of the image
xtrain = xtrain / 255.0
xtest = xtest / 255.0
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(4,3))
plt.imshow(xtest[500])
plt.show()
print(ytest[1200])
```

```python
#importing the neural network model- sequential(linear stack model)
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.utils import to_categorical
```

```python
ytrain = to_categorical(ytrain,num_classes=10)
ytest = to_categorical(ytest,num_classes=10)
```

```python
from matplotlib.cbook import flatten
model = Sequential((
    Flatten(input_shape=(28,28)),
    Dense(32, activation='relu'),
    Dense(64, activation='relu'),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
))
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(),  # Corrected spelling and instantiation
    loss=tf.keras.losses.CategoricalCrossentropy(),
    metrics=['accuracy']
)
```

```python
model.fit(
    xtrain,
    ytrain,
    epochs=10,
    batch_size=32,
    validation_data=(xtest, ytest)
)
```

```python
import numpy as np
def predict_mnist_digit(index):
  image = xtest[index]
  label = ytest[index]

  image = image.reshape(1,28,28)
  image = image / 255.0

  prediction = model.predict(image)
  predicted_label = np.argmax(prediction)


  plt.imshow(xtest[index], cmap=plt.cm.binary)
  plt.title(f'predicted MNIST digit: {predicted_label} :: actual mnist digit: {np.argmax(label)}')
  plt.show()
```
